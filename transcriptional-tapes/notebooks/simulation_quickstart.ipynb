# transcriptional-tapes quickstart

Simulate an ensemble of MS2 fluorescence traces from continuous-time promoter switching.

```python
import numpy as np
from transcriptional_tapes.model import SimulationConfig
from transcriptional_tapes.presets import three_state_default
from transcriptional_tapes.batch import simulate_dataset
from transcriptional_tapes.plotting import plot_single_trace, plot_ensemble_mean
```

```python
model = three_state_default()
config = SimulationConfig(seq_length=120, delta_t=20.0, memory_steps=7, alpha=30/140 * 7)
dataset = simulate_dataset(model=model, config=config, n_traces=40, seed=11)
time_axis = np.arange(1, config.seq_length+1) * config.delta_t
```

```python
trace0 = dataset.traces[0]
plot_single_trace(time_axis, trace0.fluo_ms2, trace0.fluo)
```

```python
plot_ensemble_mean(time_axis, dataset.observed_fluo)
```
